## Prompt 2: The five biggest weaknesses of the "obvious" approach

1. **Using K = σ√(πa) with a nominal stress.** Plugging P/(BW) or an uncorrected bending stress into √(πa) misses the geometry function. Use the specimen's own K expression.
2. **Mixing up the two forms of the equation.** Some references write K = PS/(BW^(3/2))·f(a/W), and others write K = P/(B√W)·f'(a/W) with S/W already folded into f'. Mixing the prefactor from one form with the f from the other is off by a factor of 4.
3. **Using the formula outside its validity.** This f(a/W) is fitted for **S/W = 4**. A different span ratio needs a different function.
4. **Mixing units.** Use MN and m to get MPa√m, or lb and in to get psi√in. P in kN with dimensions in mm gives a result off by orders of magnitude.
5. **Not sanity-checking the answer.** A quick check with beam bending stress (below) catches most errors. Skipping it is how a wrong prefactor goes unnoticed.

## Prompt 3: Fully worked solution

### Work backward from the goal
K_I = (load and geometry prefactor) × f(a/W). You need the right expression, f evaluated at a/W = 0.2, and consistent units.

### Step 1: The three-point bend K expression (S/W = 4)

$$K_I = \frac{PS}{BW^{3/2}}\; f\!\left(\frac{a}{W}\right)$$

$$f(x) = \frac{3\sqrt{x}\,\left[1.99 - x(1-x)(2.15 - 3.93x + 2.7x^2)\right]}{2(1+2x)(1-x)^{3/2}}$$

### Step 2: Evaluate f at x = a/W = 0.2

| Term | Value |
|---|---|
| x(1 − x) | 0.16 |
| 2.15 − 3.93(0.2) + 2.7(0.04) | 1.472 |
| Bracket: 1.99 − 0.16 × 1.472 | 1.7545 |
| Numerator: 3 × √0.2 × 1.7545 | 2.354 |
| Denominator: 2 × 1.4 × 0.8^1.5 | 2.0035 |
| **f(0.2)** | **1.175** |

### Step 3: Evaluate the prefactor (SI units: MN and m)
- PS = 0.035 × 0.203 = 7.105 × 10⁻³ MN·m
- BW^(3/2) = 0.025 × 0.0508^1.5 = 2.862 × 10⁻⁴ m^(5/2)
- PS/(BW^(3/2)) = 24.82 MPa√m

### Step 4: Result
$$K_I = 24.82 \times 1.175 = \boxed{29.2\ \text{MPa}\sqrt{\text{m}}}$$

**In US units:** 7870 × 8 / (1 × 2^1.5) × 1.175 = **26.2 ksi√in**.

Converting 26.2 ksi√in gives 28.7 MPa√m, not 29.2. The difference comes entirely from the book's B = 25 mm versus 1.0 in = 25.4 mm. With B = 25.4 mm, the SI answer is also 28.7 MPa√m.

### Sanity check with beam theory
- The nominal bending stress at midspan is σ_b = 3PS/(2BW²) = **165 MPa**.
- The crack length is a = 10.16 mm, so σ_b√(πa) = 165 × 0.179 = 29.5 MPa√m.
- The implied geometry factor is 29.2 / 29.5 = **0.99**. That's close to 1 and consistent with handbook values (Tada) for a shallow edge crack in three-point bending. The answer is reasonable.

## Prompt 4: Expert evaluation

**Beginners misunderstand:**
- **They take the geometry function on faith.** It accounts for the free edge (which raises K, like the 1.12 factor), for bending (stress falls toward the neutral axis, which lowers K as the crack deepens), and for local effects at the loading point. At a/W = 0.2 these roughly cancel, which is why the factor comes out near 1.0. At a/W = 0.5 they don't: f rises to about 2.66.
- **They treat a/W = 0.2 as a typical test value.** It isn't. Standard toughness tests (E399/E1820) require **0.45 ≤ a/W ≤ 0.55**. Shallow cracks like this one have low crack-tip constraint, which is a topic in Anderson Chapter 3.

**What professionals do differently:**
- **Check validity before trusting K.** K is only meaningful for small-scale yielding. For a K_Ic test, E399 requires B, a and (W − a) ≥ 2.5(K/σ_YS)². For a steel with σ_YS = 500 MPa, 2.5(29.2/500)² = 8.5 mm, which this specimen meets. For a lower-strength material it may not.
- **Know the sensitivity of K to crack length.** Near a/W = 0.5, a 1% error in measuring a changes K by about 2–3%. That's why test standards require crack-length measurements at several points across the thickness.
- **Choose a specimen for a reason.** SE(B) specimens are cheap and easy to machine, but use more material per test than compact specimens, and the bending loads them differently. The choice is an engineering decision, not a default.

**How to think at an advanced level:** every K solution has the same structure:

> K = (characteristic stress) × √(characteristic length) × (dimensionless geometry function)

Identify each part. Here the characteristic stress is effectively PS/(BW²) (the bending stress) and the length is W. Then check that the geometry function behaves sensibly in its limits: about 1.1 for a very shallow edge crack, and rising sharply as a/W → 1.
