# 10.1 Prompt 3

##Problem 10.1, worked: semicircular surface flaw under the Paris law

##Working backward from the answer

You need **N**, the number of cycles to grow the crack from a₀ to a_f, for 4 combinations. Getting there takes four pieces:

1. **N = ∫ dN**. Get this by rearranging the Paris law so that dN = da / (C·ΔKᵐ).
2. **ΔK as a function of a**. This is the only modeling decision: the geometry factor for a semicircular surface flaw.
3. **A closed-form integral**. For m = 3 it integrates by the power rule, so no numerical work is needed.
4. **Units**. a must be in **meters**, because C was fit with ΔK in MPa√m and da/dN in m/cycle.

The efficient approach is to derive one general formula and then substitute the four cases into it. Don't integrate four separate times.

## Step 1: Choose ΔK (the key decision)

Anderson (Ch. 2) builds the semicircular surface flaw from two corrections:
- **2/π** comes from the penny-shaped (embedded circular) crack: K = (2/π)σ√(πa).
- **1.12** is the free-surface correction for a crack breaking the surface.

Together they give:

$$\Delta K = 1.12\cdot\frac{2}{\pi}\,\Delta\sigma\sqrt{\pi a} = 0.713\,\Delta\sigma\sqrt{\pi a}$$

The problem says the crack is "small compared with the cross section," so no finite-width correction is needed. I also assume the crack stays semicircular as it grows, which the problem implies.

## Step 2: Set up and integrate

From da/dN = C(ΔK)ᵐ:

$$N = \int_{a_0}^{a_f}\frac{da}{C\,(Y\Delta\sigma\sqrt{\pi})^3\,a^{3/2}}$$

Using ∫a⁻³ᐟ² da = −2a⁻¹ᐟ²:

$$\boxed{N = \frac{2}{C\,(Y\Delta\sigma)^3\,\pi^{3/2}}\left(\frac{1}{\sqrt{a_0}}-\frac{1}{\sqrt{a_f}}\right)}$$

## Step 3: Evaluate the constant once

- (YΔσ)³ = (0.713 × 200)³ = (142.6)³ = 2.900 × 10⁶
- π³ᐟ² = 5.568
- C(YΔσ)³π³ᐟ² = 6.87×10⁻¹² × 2.900×10⁶ × 5.568 = 1.109 × 10⁻⁴

So the prefactor is **2 / 1.109×10⁻⁴ = 18,030 m^½**.

## Step 4: Substitute the four cases (a in meters)

| a (mm) | a (m) | 1/√a (m⁻½) |
|---|---|---|
| 1 | 0.001 | 31.62 |
| 2 | 0.002 | 22.36 |
| 10 | 0.010 | 10.00 |
| 20 | 0.020 | 7.07 |

| a₀ → a_f | (1/√a₀ − 1/√a_f) | **N (cycles)** |
|---|---|---|
| 1 → 10 mm | 21.62 | **≈ 3.90 × 10⁵** |
| 1 → 20 mm | 24.55 | **≈ 4.43 × 10⁵** |
| 2 → 10 mm | 12.36 | **≈ 2.23 × 10⁵** |
| 2 → 20 mm | 15.29 | **≈ 2.76 × 10⁵** |

## Step 5: Interpret the results (likely to come up in discussion)

- **The initial crack size dominates.** Doubling a₀ from 1 to 2 mm cuts life by about 43%. Doubling a_f from 10 to 20 mm adds only about 14–24%. Most of the life is spent while the crack is small, because da/dN ∝ a³ᐟ². This is why inspection resolution (the smallest crack you can detect, a₀) matters more than the exact critical size.
- **The answer is very sensitive to Y,** because N ∝ 1/Y³. If your course uses the Newman–Raju solution instead (Y ≈ 0.66 at the deepest point), every N is about 25% higher: 4.86, 5.52, 2.78 and 3.44 × 10⁵ cycles. If you're matching an answer key, check which ΔK expression your instructor used.

# Problem 10.1 from an expert's view: beyond the textbook answer

I haven't seen your own work, so I'm assessing the approach we used: integrate the Paris law in closed form with ΔK = 0.713 Δσ√(πa). **For a textbook problem it's correct and efficient.** An expert would treat it as a first-order estimate, though, because several simplifications are built into it. Here is where those are and how to think past them.

---

## 1. What beginners misunderstand

**"The answer is a number."** It's really a model output that depends on Y³, C and Δσ³. A 10% error in stress changes life by about 25–37%. The scatter in C for the same material is often a factor of 2–3 in life. Professionals report N as a range, or as a design value with explicit margins, never as "389,830 cycles."

**"Y is a constant."** It isn't, even for this "simple" flaw. For a semicircle, the Newman–Raju solution gives a higher K at the **surface point** (≈1.14·√(πa/Q)) than at the **deepest point** (≈1.04·√(πa/Q)). So the crack grows faster along the surface, flattens into a semi-ellipse, and its aspect ratio drifts as it grows. The assumption that the crack "stays semicircular" is convenient, not physical.

**"Units are bookkeeping."** C is not a material constant in any physical sense. It's a curve-fit coefficient tied to specific units. Put a in mm into this C and you're off by a factor of 1000^(3/2)/1000, about 31.6×, with no warning.

**"Paris law covers the whole life."** It only describes Region II, the straight line on a log-log plot. Check the ΔK range this problem actually covers:

| a | ΔK (MPa√m) |
|---|---|
| 1 mm | ≈ 8.0 |
| 2 mm | ≈ 11.3 |
| 10 mm | ≈ 25.3 |
| 20 mm | ≈ 35.7 |

For many steels, ΔK_th is about 3–8 MPa√m. **The 1 mm start is near threshold**, which is where Paris law is least accurate and where most of the life is spent.

---

## 2. Where the life actually goes (the most important insight)

Take the 1 → 20 mm case. The fraction of total life spent growing from 1 to 2 mm is:

(31.62 − 22.36) / (31.62 − 7.07) ≈ **38%**

Growing from 1 to 10 mm takes about 88% of the total, while the second half of the crack length, 10 to 20 mm, takes only about 12%.

What follows from this:
- **N depends mostly on a₀, while a_f barely matters.** This is why the critical crack size can be fairly rough, but the initial flaw size can't be.
- The least reliable part of the model (near threshold, possibly small-crack effects, uncertain Y) controls most of the answer.
- So an expert spends effort on **defining a₀**, not on refining a_f.

---

## 3. What experienced practitioners do differently

**They define a₀ by inspection capability, not by assumption.** a₀ is set from the probability of detection of the NDI method, typically a90/95: the crack size detected 90% of the time with 95% confidence. "Assume 1 mm" turns into "UT reliably finds 1.3 mm in this geometry, so a₀ = 1.3 mm."

**They bound a_f with real failure criteria.** Using the 1→20 mm case:
- Fracture: K_max = K_IC. At R = 0 and a = 20 mm, K_max ≈ 36 MPa√m. That's fine for most steels but marginal for high-strength aluminum (K_IC ≈ 25–30 MPa√m), where the crack **would fracture before reaching 20 mm**. The textbook skips this check.
- Net-section yield and leak-before-break (through-wall in a pressure vessel) are also possible limits.
- a_f is whichever of these happens first.

**They use a growth law with mean-stress and threshold effects.** The problem gives Δσ but no R-ratio, and crack growth depends strongly on R. Practitioners use Walker, Forman or the NASGRO equation, which include R, ΔK_th and the acceleration as K_max approaches K_c.

**They integrate numerically with an evolving geometry.** Tools such as NASGRO, AFGROW and FRANC3D update the crack shape cycle by cycle, track both the deepest and surface points, and account for:
- Load sequence and retardation: overloads slow growth, so constant-amplitude results are usually conservative.
- Residual stresses from welds or shot peening.
- Environment: corrosion fatigue can raise da/dN by 10× or more.

**They turn life into an inspection plan.** The question that matters is how often to inspect, not how long it lasts. Typically the inspection interval is N/2 or N/3, so a crack gets at least two chances to be found before it becomes critical. In the 2→20 mm case, N ≈ 2.8 × 10⁵ cycles gives an interval of about 1.4 × 10⁵ cycles.

**They check LEFM validity.** Δσ = 200 MPa is moderate. Experts still compare it with yield strength and check the plastic-zone size (≈ (1/2π)(K/σ_y)²) against the crack size and ligament. For low-yield materials at the larger cracks, LEFM starts to lose validity.

---

## 4. How to think at an advanced level

**Ask what controls the answer before computing it.** Here, N ∝ a₀^(−1/2) · Y⁻³ · Δσ⁻³ · C⁻¹. Write out these sensitivities first, then spend your effort on whichever input is most uncertain and has the largest exponent.

**Run a sanity check:**
- Dimensions: [m^½] / ([m/cycle]/[MPa√m]³ · [MPa]³ · [m^(3/2)]) gives cycles.
- Order of magnitude: about 10⁵ cycles for mm-scale cracks at 200 MPa is typical for steel.
- Limits: if a_f → ∞, N stays finite. That's a signature of m > 2.

**Bracket the result instead of reporting one number.** Use Y between 0.66 and 0.73 and an upper-bound C:

| Case | N range (cycles) |
|---|---|
| 1 → 10 mm | ≈ 2–5 × 10⁵ |
| 2 → 20 mm | ≈ 1.4–3.5 × 10⁵ |

Make decisions on the conservative end.

**Name every assumption out loud:**
- Constant-amplitude loading
- R unspecified (Paris law fitted at one R)
- Semicircular shape maintained
- Deepest point controls growth
- No threshold
- No environment
- Infinite body
- LEFM valid

Then state which assumptions make the answer conservative and which make it unconservative. Here, ignoring threshold and assuming constant amplitude are conservative. Ignoring a K_IC check and environment can be unconservative.

**Question the question.** "How many cycles?" usually stands in for "Is this part safe until the next inspection, and if not, what do we change?" Your levers are:
- **Better NDI**, which lowers a₀ and has the biggest effect.
- **Lower stress**, which pays off as a cube.
- **A tougher material**, which raises a_f.
- **Compressive residual stress**, which effectively lowers ΔK.

## Bottom line

Your approach is the right closed-form estimate and earns full marks on the problem. To think like a practitioner, add three things:
1. **Treat a₀ as the controlling input.** About 38% of the life is spent doubling the crack from 1 to 2 mm.
2. **Check that Paris law and the failure criterion actually hold** across the ΔK range, from about 8 to 36 MPa√m here.
3. **Report a bounded life tied to an inspection interval** rather than one precise number.

# Your biggest leverage point: understanding ΔK = YΔσ√(πa)

**The one thing to master is the stress intensity factor**, specifically why ΔK = Y·Δσ·√(πa) and what each term means. Everything else in this problem is either given or routine calculus. If you truly understand K, you can solve this problem, every variant of it, and most of the rest of Anderson.

## Why this is the leverage point

The problem is a chain of three links, and only one requires understanding:

| Link | What it is | Difficulty |
|---|---|---|
| σ → **K** | Turning applied stress plus crack geometry into crack-tip severity | **Conceptual: this is the whole subject** |
| K → da/dN | Paris law, given to you as an empirical fit | Plug-in |
| da/dN → N | Separate the variables and integrate a power law | First-year calculus |

All the judgment in the problem lives in the first link:
- Choosing Y (0.713 versus 0.66) is a K question.
- "Why does a semicircle turn into an ellipse?" is a K question.
- "Why is most of the life spent while the crack is small?" is a K question too, since ΔK ∝ √a.
- "Will it fracture before 20 mm?" is a K question (K_max vs. K_IC).

People get stuck on fatigue problems because K is still a formula they plug into rather than an idea they understand. The integration is rarely the problem.

## What "understanding K" actually means

You understand it when you can explain each piece without notes:

1. **√a:** The stress near a crack tip falls off as σ_ij ∝ K/√(2πr). K measures how strong that singular field is, and a longer crack concentrates load over a longer lever, so K grows as √a.
2. **π:** This comes from the Griffith/Irwin through-crack solution in an infinite plate, the reference case where Y = 1.
3. **Y:** Every other geometry is the reference case times a correction:
   - **2/π** because a crack embedded as a circle (penny shape) is surrounded by material, which holds its faces together better than a through-crack's.
   - **1.12** because a free surface lets the crack mouth open more, which raises K.
4. **ΔK rather than K:** Fatigue is driven by the *range* of crack-tip loading per cycle, so Δσ replaces σ.

## What follows once you have it

With K understood, the rest of the problem is derivation, not memorization:
- da/dN = C(YΔσ√(πa))³ ∝ a^1.5, so the crack accelerates as it grows.
- Integrating ∫a^(−1.5) da gives the 1/√a₀ − 1/√a_f form, and the 1/√a₀ term dominates.
- So the initial flaw size controls the life. That's the engineering lesson of this chapter, and it falls straight out of the √a.

## Most efficient path to complete understanding

1. **Anderson Ch. 2, sections on K and the Griffith/Irwin through-crack.** Focus on the through-crack in an infinite plate and the K-based fracture criterion.
2. **Anderson Ch. 2, geometry corrections.** Work out the penny crack and the surface-crack correction, and see where 2/π and 1.12 come from.
3. **Re-derive the Problem 10.1 formula from a blank page** without looking. If you can, you own it.
4. **Test yourself with three predictions before calculating:**
   - What happens to N if Δσ doubles? (It drops by 8×.)
   - What happens if a₀ is halved? (N grows, but by less than 2×.)
   - Why does a_f barely matter?

   If you can answer these from the equation's structure, your understanding is complete.

Steps 3 and 4 are the milestones. Once you can derive the formula unaided and predict those sensitivities without calculating, you have both the solution and the mechanics behind it.
