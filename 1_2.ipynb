# Problem 1.2: same five-step protocol

**Answer: half-length a_c ≈ 79.6 mm, so the total critical crack length is 2a_c ≈ 159 mm (6.27 in).**

---

## Prompt 1: What you're actually solving, and the three key questions

**What you're solving.** This is the most basic fracture calculation there is. A through-thickness center crack in a wide plate fails when K_I = σ√(πa) reaches K_Ic. You rearrange that for crack size. It's the first appearance of the idea that runs through all your other problems: **fracture happens when the crack-driving force equals the material's resistance**.

**The three questions I'd ask, and the answers I'm assuming:**
1. *Is the plate infinitely wide?* Figure 1.8 shows a through crack in a wide plate, so I use K = σ√(πa) with no finite-width correction.
2. *Does "crack length" mean a or 2a?* For a center crack, **a is the half-length**. I give both, because this is the most common place to lose marks.
3. *SI or US units?* I give both. Here the book's two sets of numbers are consistent.

---

## Prompt 2: The five biggest weaknesses of the "obvious" approach

1. **Reporting a as the crack length.** The formula gives the half-length. The physical crack is 2a = 159 mm. Reporting 79.6 mm as "the crack length" is off by 2×.
2. **Adding the 1.12 edge-crack factor.** That factor applies to an edge crack. A center crack in a wide plate has a geometry factor of 1.
3. **Assuming the plate is infinitely wide.** A 159 mm crack is large. If the plate is, say, 300 mm wide, the finite-width correction is significant and the true critical size is smaller.
4. **Ignoring plasticity.** "Assuming linear elastic" hides an assumption: that the plastic zone is small. That depends on yield strength, which isn't given.
5. **Treating a_c as an allowable flaw size.** a_c is the size at failure, not a safe size. Design uses a safety factor and inspection, as you saw in 10.2.

## Prompt 3: Fully worked solution

### Work backward from the goal
You want a_c. Fracture occurs when K_I = K_Ic. For this geometry K_I = σ√(πa), so solve for a.

### Step 1: The fracture criterion
$$\sigma\sqrt{\pi a_c} = K_{Ic}$$

### Step 2: Rearrange
$$a_c = \frac{1}{\pi}\left(\frac{K_{Ic}}{\sigma}\right)^2$$

### Step 3: Substitute (SI)
$$a_c = \frac{1}{\pi}\left(\frac{50.0}{100}\right)^2 = \frac{0.25}{\pi} = 0.0796\ \text{m}$$

- **a_c = 79.6 mm** (half-length)
- **2a_c = 159 mm** (total crack length)

### Step 4: Check in US units
$$a_c = \frac{1}{\pi}\left(\frac{45.5}{14.5}\right)^2 = \frac{9.846}{\pi} = 3.13\ \text{in}$$

So 2a_c = **6.27 in**, which equals 159 mm and confirms the SI result.

## Prompt 4: Expert evaluation

**Beginners misunderstand:**
- **How strongly crack size depends on stress.** a_c ∝ (K_Ic/σ)². Doubling the stress cuts the critical crack size by 4×, to about 40 mm total. That's why fracture is so sensitive to overloads.
- **That K_Ic is a material property while K_I is a loading quantity.** The criterion compares the driving force (geometry plus load) with the material's resistance. Every fracture problem in your set has this structure.
- **That "linear elastic" is a real assumption.** It needs checking, not just stating.

**What professionals do differently:**
- **Apply a finite-width correction.** For a plate of width W, K = σ√(πa)·√(sec(πa/W)). If W = 500 mm, the correction is about 5% and a_c drops by roughly 10%. For narrower plates it's larger, and it requires iteration because a_c appears on both sides.
- **Check plasticity with the yield strength.** The Irwin plastic zone is r_y = (1/2π)(K_Ic/σ_YS)².
  - For σ_YS = 500 MPa: r_y ≈ 1.6 mm, much smaller than a = 80 mm, so LEFM is fine.
  - For σ_YS = 200 MPa, two things change. The applied stress is half of yield, so net-section yielding may govern. And r_y ≈ 10 mm is large enough to need a plasticity correction or a two-criteria approach (the failure assessment diagram, covered later in Anderson).
- **Check whether the material is thick enough for K_Ic to apply.** K_Ic is plane-strain toughness. A thin plate has a higher apparent toughness, so using K_Ic is conservative but may be very conservative.
- **Turn a_c into decisions.** Compare a_c with the smallest crack inspection can detect. A 159 mm critical crack in a wide plate is easy to find by visual inspection, so this is a forgiving design. Compare that with Problem 10.2, where a_c was about 9 mm.

**How to think at an advanced level:** see that this problem and every other one in your set share a single fracture equation:

> K_applied(σ, a, geometry) = K_material

- **1.2:** solve for **a** (critical crack size)
- **10.2:** used it to find the endpoint for fatigue
- **2.11:** evaluated the left side for a specific geometry
- **Fatigue (10.1–10.5):** cycle the left side and grow the crack until the two sides meet

You can always solve for any one of σ, a, or K_Ic given the other two.
